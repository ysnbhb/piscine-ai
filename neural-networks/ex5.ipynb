```python
import numpy as np
import math
```

```python
class Neuron:
    def __init__(self, weight1, weight2, bias, regression):
        self.weights_1 = weight1
        self.weights_2 = weight2
        self.bias = bias
        self.rg = regression

    def feedforward(self, x1, x2):
        y = x1 * self.weights_1 + x2 * self.weights_2 + self.bias 
        if not self.rg :
            y =  1 / (1 + np.exp(-y))
        return y
```

```python
neuron = Neuron(0,1,4, True)
neuron.feedforward(2,3)
```

```python
class OurNeuralNetwork:

    def __init__(self, neuron_h1, neuron_h2, neuron_o1):
        self.h1 = neuron_h1
        self.h2 = neuron_h2
        self.o1 = neuron_o1

    def feedforward(self, x1, x2):
        h1_out = self.h1.feedforward(x1 , x2)
        h2_out = self.h2.feedforward(x1 , x2)
        return self.o1.feedforward(h1_out , h2_out)
regression =True
neuron_h1 = Neuron(0.05, 0.001, 0, False)
neuron_h2 = Neuron(0.002, 0.003, 0, False)
neuron_o1 = Neuron(2,7,10, regression)
```

```python
network = OurNeuralNetwork(neuron_h1 , neuron_h2 , neuron_o1)

students = {
    "Bob": (12, 15 , 16),
    "Eli": (10, 9 , 10),
    "Tom": (18, 18 , 19),
    "Ryan": (13, 14 , 16)
}

y_true = []
y_pred = []

for name, (math_score, chemistry_score, physics_score) in students.items():
    output = network.feedforward(math_score, chemistry_score)
    y_pred.append(output)
    y_true.append(physics_score)
    print(f"{name}: {output}")
y_pred  = np.array(y_pred)
y_true  = np.array(y_true)
```

```python
y_pred = np.array(y_pred)
y_true = np.array(y_true)

mse = np.mean((y_true - y_pred)**2)
print(f"MSE: {mse}")
```
