```python
import numpy as np 
from sklearn.metrics import log_loss
```

```python
def log_loss_custom(y_true  , y_pre):
    y= (y_true * np.log(y_pre) + (1 - y_true) * np.log(1-y_pre))
    return  -(1/len(y_true) * np.sum(y))
```

```python
y_true = np.array([0,1,1,0,1])
y_pred = np.array([0.1,0.8,0.6, 0.5, 0.3])
```

```python
log_loss_custom(y_true , y_pred)
```

```python
log_loss(y_true , y_pred)
```
