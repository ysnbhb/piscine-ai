```python
import math 
import numpy as np 
```

```python
class Neuron:
    def __init__(self, weight1, weight2, bias):
        self.weights_1 = weight1
        self.weights_2 = weight2
        self.bias = bias

    def feedforward(self, x1, x2):
        y = (x1* self.weights_1 + x2 * self.weights_2) + self.bias

        return  1 / (1 + math.exp(-y))
```

```python
class OurNeuralNetwork:

    def __init__(self, neuron_h1, neuron_h2, neuron_o1):
        self.h1 = neuron_h1
        self.h2 = neuron_h2
        self.o1 = neuron_o1

    def feedforward(self, x1, x2):
        h1_out = self.h1.feedforward(x1 , x2)
        h2_out = self.h2.feedforward(x1 , x2)
        return self.o1.feedforward(h1_out , h2_out)
```

```python
def log_loss_custom(y_true  , y_pre):
    y= (y_true * np.log(y_pre) + (1 - y_true) * np.log(1-y_pre))
    return  -(1/len(y_true) * np.sum(y))
```

```python
neuron_h1 = Neuron(0.05, 0.001, 0)
neuron_h2 = Neuron(0.02, 0.003, 0)
neuron_o1 = Neuron(2,0,0)
```

```python
network = OurNeuralNetwork(neuron_h1 , neuron_h2 , neuron_o1)
```

```python
students = {
    "Bob": (12, 15 , 1),
    "Eli": (10, 9 , 0),
    "Tom": (18, 18 , 1),
    "Ryan": (13, 14 , 1)
}

y_true = []
y_pre = []

for name, (math_score, chemistry_score , sus) in students.items():
    output = network.feedforward(math_score, chemistry_score)
    y_pre.append(output)
    y_true.append(sus)
    print(f"{name}: {output}")
y_pred  = np.array([i  for i in y_pre])
y_true  = np.array([i  for i in y_true])
```

```python
log_loss_custom(y_true , y_pred)
```
